```python
from langchain.chat_models import init_chat_model
from pydantic import BaseModel, Field

qwen_35 = init_chat_model(
    model="qwen_35",
    model_provider="openai",
    temperature=0.6,
    top_p=0.95,
    max_retries=2,
    api_key="not-needed",
    base_url="http://localhost:8086/v1",
    extra_body={
        "top_k": 20,
        "chat_template_kwargs": {"enable_thinking": False},
    },
)

class JobMarketDS(BaseModel):
    company: str = Field(
        description="name of the company"
    )
    role: str = Field(
        description="title of the role role of the openning"
    )
    problem: str = Field(
        description="problem the company is trying to solve/service they are trying to provide"
    )
    skills: str = Field(
        description="the skills that the job requires, like frameworks"
    )
    pain: str = Field(
        description="in the process of problem solving what is their main obstacle"
    )
qwen_35_with_structured_output = qwen_35.with_structured_output(JobMarketDS)
```
